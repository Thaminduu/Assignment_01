```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Load the image (BGR format by default in OpenCV)
img_bgr = cv2.imread('C:\\Users\\thami\\Downloads\\images\\images\\q3.jpeg')

# 2. Convert from BGR to CIELAB color space
img_lab = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2LAB)

# 3. Split into L*, a*, and b* channels
# In OpenCV 8-bit representation, L ranges from 0 to 255 (scaled from standard 0-100)
L, a, b = cv2.split(img_lab)

# 4. Define the gamma value
# gamma < 1.0 brightens mid-tones (e.g., 0.5 - 0.7)
# gamma > 1.0 darkens mid-tones (e.g., 1.5 - 2.2)
gamma = 0.6  # <-- State your chosen gamma value here

# 5. Apply power-law (gamma) transformation to the L channel
# Normalize L to [0, 1], apply L_out = L_in ^ gamma, then rescale back to [0, 255]
L_norm = L.astype(np.float32) / 255.0
L_gamma = np.power(L_norm, gamma)
L_corrected = np.clip(L_gamma * 255.0, 0, 255).astype(np.uint8)

# 6. Merge the modified L channel back with the original a* and b* channels
img_lab_corrected = cv2.merge([L_corrected, a, b])

# 7. Convert back to BGR color space for display/saving
img_bgr_corrected = cv2.cvtColor(img_lab_corrected, cv2.COLOR_LAB2BGR)

# Save result
cv2.imwrite('gamma_corrected.jpg', img_bgr_corrected)


plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB))
plt.title('Original Image')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(img_bgr_corrected, cv2.COLOR_BGR2RGB))
plt.title('Corrected Image')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Create a mock image (random colors)

img = cv2.imread('C:\\Users\\thami\\Downloads\\images\\images\\q4.jpeg')

# Convert RGB → HSV
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# Split channels
H, S, V = cv2.split(hsv)

# Display channels
plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(H, cmap='hsv')
plt.title("Hue")

plt.subplot(1,3,2)
plt.imshow(S, cmap='gray')
plt.title("Saturation")

plt.subplot(1,3,3)
plt.imshow(V, cmap='gray')
plt.title("Value")

plt.show()
```

```python
# Parameters
a = 0.5
sigma = 70

# Convert S to float for processing
x = S.astype(np.float32)

# Apply transformation
S_transformed = x + a * 128 * np.exp(-((x - 128) ** 2) / (2 * sigma ** 2))

# Clip to valid range
S_transformed = np.clip(S_transformed, 0, 255).astype(np.uint8)

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(H, cmap='hsv')
plt.title("Hue")

plt.subplot(1,3,2)
plt.imshow(S_transformed, cmap='gray')
plt.title("Saturation")

plt.subplot(1,3,3)
plt.imshow(V, cmap='gray')
plt.title("Value")

plt.show()
```

```python
a_values = [0.3, 0.5, 0.7]

plt.figure(figsize=(12,4))

for i, a in enumerate(a_values):
    S_temp = x + a * 128 * np.exp(-((x - 128) ** 2) / (2 * sigma ** 2))
    S_temp = np.clip(S_temp, 0, 255).astype(np.uint8)
    
    hsv_temp = cv2.merge([H, S_temp, V])
    img_temp = cv2.cvtColor(hsv_temp, cv2.COLOR_HSV2RGB)
    
    plt.subplot(1,3,i+1)
    plt.imshow(img_temp)
    plt.title(f"a = {a}")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


img = cv2.imread('C:\\Users\\thami\\Downloads\\images\\images\\q5.jpeg', cv2.IMREAD_GRAYSCALE)

def histogram_equalization(image):
    # Flatten image to 1D array
    flat = image.flatten()
    
    # Step 1: Compute histogram
    hist = np.zeros(256, dtype=int)
    for pixel in flat:
        hist[pixel] += 1
    
    # Step 2: Normalize histogram (PDF)
    pdf = hist / len(flat)
    
    # Step 3: Compute CDF
    cdf = np.cumsum(pdf)
    
    # Step 4: Create transformation function
    transform = np.round(cdf * 255).astype(np.uint8)
    
    # Step 5: Map original image using transform
    equalized = transform[image]
    
    return equalized, hist, transform

image_path = r"C:\Users\thami\Downloads\images\images\q5.jpg"

img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

equalized_img, hist_before, transform = histogram_equalization(img)
hist_after = np.zeros(256, dtype=int)
for pixel in equalized_img.flatten():
    hist_after[pixel] += 1

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")

plt.subplot(1,2,2)
plt.imshow(equalized_img, cmap='gray')
plt.title("Equalized Image")

plt.show()
```

```python
plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.bar(range(256), hist_before)
plt.title("Histogram Before Equalization")

plt.subplot(1,2,2)
plt.bar(range(256), hist_after)
plt.title("Histogram After Equalization")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


img = cv2.imread('C:\\Users\\thami\\Downloads\\images\\images\\q6.jpeg')

# Convert to HSV
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
H, S, V = cv2.split(hsv)

# Display planes in grayscale
plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(H, cmap='gray')
plt.title("Hue")

plt.subplot(1,3,2)
plt.imshow(S, cmap='gray')
plt.title("Saturation")

plt.subplot(1,3,3)
plt.imshow(V, cmap='gray')
plt.title("Value")

plt.show()
```

```python
# Threshold on Value channel
_, mask = cv2.threshold(V, 120, 255, cv2.THRESH_BINARY)

# Display mask
plt.imshow(mask, cmap='gray')
plt.title("Foreground Mask")
plt.show()
```

```python
# Extract foreground using mask
foreground = cv2.bitwise_and(V, V, mask=mask)

# Compute histogram
hist = np.zeros(256, dtype=int)
for pixel in foreground.flatten():
    hist[pixel] += 1

# Display histogram
plt.bar(range(256), hist)
plt.title("Foreground Histogram")
plt.show()
```

```python
cdf = np.cumsum(hist)

# Normalize CDF for visualization (optional)
cdf_normalized = cdf / cdf.max()

plt.plot(cdf_normalized)
plt.title("CDF of Foreground")
plt.show()
```

```python
# Avoid division by zero (ignore zero values in mask)
cdf_min = cdf[np.nonzero(cdf)].min()

# Histogram equalization formula
cdf_eq = (cdf - cdf_min) * 255 / (cdf.max() - cdf_min)
cdf_eq = np.clip(cdf_eq, 0, 255).astype(np.uint8)

# Apply mapping only to foreground
equalized_foreground = cdf_eq[foreground]
```

```python
# Background (inverse mask)
background = cv2.bitwise_and(V, V, mask=cv2.bitwise_not(mask))

# Combine equalized foreground + original background
combined_V = cv2.add(equalized_foreground, background)

# Replace V channel
hsv_result = cv2.merge([H, S, combined_V])

# Convert back to RGB
result = cv2.cvtColor(hsv_result, cv2.COLOR_HSV2RGB)
```

```python
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(12,8))

plt.subplot(2,3,1)
plt.imshow(img)
plt.title("Original")

plt.subplot(2,3,2)
plt.imshow(H, cmap='gray')
plt.title("Hue")

plt.subplot(2,3,3)
plt.imshow(S, cmap='gray')
plt.title("Saturation")

plt.subplot(2,3,4)
plt.imshow(V, cmap='gray')
plt.title("Value")

plt.subplot(2,3,5)
plt.imshow(mask, cmap='gray')
plt.title("Mask")

plt.subplot(2,3,6)
plt.imshow(result)
plt.title("Final Result")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Mock grayscale image (replace with Fig.7 if available)
img = cv2.imread('C:\\Users\\thami\\Downloads\\images\\images\\q7.jpeg', cv2.IMREAD_GRAYSCALE)

plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.show()
```

```python
# Sobel kernels
sobel_x = np.array([[1, 0, -1],
                    [2, 0, -2],
                    [1, 0, -1]])

sobel_y = np.array([[1,  2,  1],
                    [0,  0,  0],
                    [-1, -2, -1]])

# Apply filters
gx = cv2.filter2D(img, -1, sobel_x)
gy = cv2.filter2D(img, -1, sobel_y)

# Gradient magnitude
grad = np.sqrt(gx.astype(float)**2 + gy.astype(float)**2)
grad = np.clip(grad, 0, 255).astype(np.uint8)

# Display
plt.imshow(grad, cmap='gray')
plt.title("Sobel using filter2D")
plt.show()
```

```python
def convolve2d(image, kernel):
    h, w = image.shape
    kh, kw = kernel.shape
    
    pad_h, pad_w = kh//2, kw//2
    padded = np.pad(image, ((pad_h, pad_h), (pad_w, pad_w)), mode='constant')
    
    output = np.zeros_like(image, dtype=float)
    
    for i in range(h):
        for j in range(w):
            region = padded[i:i+kh, j:j+kw]
            output[i, j] = np.sum(region * kernel)
    
    return output

# Apply custom convolution
gx_custom = convolve2d(img, sobel_x)
gy_custom = convolve2d(img, sobel_y)

grad_custom = np.sqrt(gx_custom**2 + gy_custom**2)
grad_custom = np.clip(grad_custom, 0, 255).astype(np.uint8)

# Display
plt.imshow(grad_custom, cmap='gray')
plt.title("Custom Sobel")
plt.show()
```

```python
# Separable kernels
kx = np.array([[1, 0, -1]])        # horizontal
ky = np.array([[1], [2], [1]])     # vertical

# First horizontal, then vertical
temp = cv2.filter2D(img, -1, kx)
gx_sep = cv2.filter2D(temp, -1, ky)

# For Y direction (transpose idea)
temp2 = cv2.filter2D(img, -1, ky.T)
gy_sep = cv2.filter2D(temp2, -1, kx.T)

# Gradient magnitude
grad_sep = np.sqrt(gx_sep.astype(float)**2 + gy_sep.astype(float)**2)
grad_sep = np.clip(grad_sep, 0, 255).astype(np.uint8)

# Display
plt.imshow(grad_sep, cmap='gray')
plt.title("Separable Sobel")
plt.show()
```

```python
plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(grad, cmap='gray')
plt.title("filter2D")

plt.subplot(1,3,2)
plt.imshow(grad_custom, cmap='gray')
plt.title("Custom")

plt.subplot(1,3,3)
plt.imshow(grad_sep, cmap='gray')
plt.title("Separable")

plt.show()
```
