```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


img = cv2.imread('C:\\Users\\thami\\Downloads\\images\\images\\q5.jpeg', cv2.IMREAD_GRAYSCALE)

def histogram_equalization(image):
    # Flatten image to 1D array
    flat = image.flatten()
    
    # Step 1: Compute histogram
    hist = np.zeros(256, dtype=int)
    for pixel in flat:
        hist[pixel] += 1
    
    # Step 2: Normalize histogram (PDF)
    pdf = hist / len(flat)
    
    # Step 3: Compute CDF
    cdf = np.cumsum(pdf)
    
    # Step 4: Create transformation function
    transform = np.round(cdf * 255).astype(np.uint8)
    
    # Step 5: Map original image using transform
    equalized = transform[image]
    
    return equalized, hist, transform

image_path = r"C:\Users\thami\Downloads\images\images\q5.jpg"

img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

equalized_img, hist_before, transform = histogram_equalization(img)
hist_after = np.zeros(256, dtype=int)
for pixel in equalized_img.flatten():
    hist_after[pixel] += 1

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")

plt.subplot(1,2,2)
plt.imshow(equalized_img, cmap='gray')
plt.title("Equalized Image")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


img = cv2.imread('C:\\Users\\thami\\Downloads\\images\\images\\q6.jpeg')

# Convert to HSV
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
H, S, V = cv2.split(hsv)

# Display planes in grayscale
plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(H, cmap='gray')
plt.title("Hue")

plt.subplot(1,3,2)
plt.imshow(S, cmap='gray')
plt.title("Saturation")

plt.subplot(1,3,3)
plt.imshow(V, cmap='gray')
plt.title("Value")

plt.show()
```

```python
# Threshold on Value channel
_, mask = cv2.threshold(V, 120, 255, cv2.THRESH_BINARY)

# Display mask
plt.imshow(mask, cmap='gray')
plt.title("Foreground Mask")
plt.figure(figsize=(4, 4))
plt.show()
```

```python
# Extract foreground using mask
foreground = cv2.bitwise_and(V, V, mask=mask)

# Compute histogram
hist = np.zeros(256, dtype=int)
for pixel in foreground.flatten():
    hist[pixel] += 1

cdf = np.cumsum(hist)

# Normalize CDF for visualization (optional)
cdf_normalized = cdf / cdf.max()
# Display histogram
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
plt.plot(cdf_normalized)
plt.title("CDF of Foreground")
plt.subplot(1, 2, 2)
plt.bar(range(256), hist)
plt.title("Foreground Histogram")
plt.show()
```

```python
# Avoid division by zero (ignore zero values in mask)
cdf_min = cdf[np.nonzero(cdf)].min()

# Histogram equalization formula
cdf_eq = (cdf - cdf_min) * 255 / (cdf.max() - cdf_min)
cdf_eq = np.clip(cdf_eq, 0, 255).astype(np.uint8)

# Apply mapping only to foreground
equalized_foreground = cdf_eq[foreground]
```

```python
# Background (inverse mask)
background = cv2.bitwise_and(V, V, mask=cv2.bitwise_not(mask))

# Combine equalized foreground + original background
combined_V = cv2.add(equalized_foreground, background)

# Replace V channel
hsv_result = cv2.merge([H, S, combined_V])

# Convert back to RGB
result = cv2.cvtColor(hsv_result, cv2.COLOR_HSV2RGB)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Mock grayscale image (replace with Fig.7 if available)
img = cv2.imread('C:\\Users\\thami\\Downloads\\images\\images\\q7.jpeg', cv2.IMREAD_GRAYSCALE)

# Sobel kernels
sobel_x = np.array([[1, 0, -1],
                    [2, 0, -2],
                    [1, 0, -1]])

sobel_y = np.array([[1,  2,  1],
                    [0,  0,  0],
                    [-1, -2, -1]])

# Apply filters
gx = cv2.filter2D(img, -1, sobel_x)
gy = cv2.filter2D(img, -1, sobel_y)

# Gradient magnitude
grad = np.sqrt(gx.astype(float)**2 + gy.astype(float)**2)
grad = np.clip(grad, 0, 255).astype(np.uint8)

# Display
plt.imshow(grad, cmap='gray')
plt.title("Sobel using filter2D")
plt.figure(figsize=(4, 4))
plt.show()
```

```python
def convolve2d(image, kernel):
    h, w = image.shape
    kh, kw = kernel.shape
    
    pad_h, pad_w = kh//2, kw//2
    padded = np.pad(image, ((pad_h, pad_h), (pad_w, pad_w)), mode='constant')
    
    output = np.zeros_like(image, dtype=float)
    
    for i in range(h):
        for j in range(w):
            region = padded[i:i+kh, j:j+kw]
            output[i, j] = np.sum(region * kernel)
    
    return output

# Apply custom convolution
gx_custom = convolve2d(img, sobel_x)
gy_custom = convolve2d(img, sobel_y)

grad_custom = np.sqrt(gx_custom**2 + gy_custom**2)
grad_custom = np.clip(grad_custom, 0, 255).astype(np.uint8)

# Display
plt.imshow(grad_custom, cmap='gray')
plt.title("Custom Sobel")
plt.figure(figsize=(4, 4))
plt.show()
```

```python
# Separable kernels
kx = np.array([[1, 0, -1]])        # horizontal
ky = np.array([[1], [2], [1]])     # vertical

# First horizontal, then vertical
temp = cv2.filter2D(img, -1, kx)
gx_sep = cv2.filter2D(temp, -1, ky)

# For Y direction (transpose idea)
temp2 = cv2.filter2D(img, -1, ky.T)
gy_sep = cv2.filter2D(temp2, -1, kx.T)

# Gradient magnitude
grad_sep = np.sqrt(gx_sep.astype(float)**2 + gy_sep.astype(float)**2)
grad_sep = np.clip(grad_sep, 0, 255).astype(np.uint8)

# Display
plt.imshow(grad_sep, cmap='gray')
plt.title("Separable Sobel")
plt.figure(figsize=(4, 4))
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# ----------------------------
# Load images
# ----------------------------
original = cv2.imread(r'C:\Users\thami\Downloads\images\images\q8.png')
small = cv2.imread(r'C:\Users\thami\Downloads\images\images\q8_small.png')

# Check loading
if original is None or small is None:
    raise ValueError("Image not loaded. Check file paths!")

# Convert to grayscale
original = cv2.cvtColor(original, cv2.COLOR_BGR2GRAY)
small = cv2.cvtColor(small, cv2.COLOR_BGR2GRAY)


# ----------------------------
# Nearest Neighbor Zoom
# ----------------------------
def zoom_nearest(image, scale):
    h, w = image.shape
    new_h, new_w = int(h * scale), int(w * scale)

    zoomed = np.zeros((new_h, new_w), dtype=image.dtype)

    for i in range(new_h):
        for j in range(new_w):
            x = min(int(i / scale), h - 1)
            y = min(int(j / scale), w - 1)
            zoomed[i, j] = image[x, y]

    return zoomed


# ----------------------------
# Bilinear Interpolation Zoom
# ----------------------------
def zoom_bilinear(image, scale):
    h, w = image.shape
    new_h, new_w = int(h * scale), int(w * scale)

    zoomed = np.zeros((new_h, new_w), dtype=np.float32)

    for i in range(new_h):
        for j in range(new_w):

            x = i / scale
            y = j / scale

            x1 = min(int(x), h - 1)
            y1 = min(int(y), w - 1)
            x2 = min(x1 + 1, h - 1)
            y2 = min(y1 + 1, w - 1)

            dx = x - x1
            dy = y - y1

            value = (
                image[x1, y1] * (1 - dx) * (1 - dy) +
                image[x2, y1] * dx * (1 - dy) +
                image[x1, y2] * (1 - dx) * dy +
                image[x2, y2] * dx * dy
            )

            zoomed[i, j] = value

    return np.clip(zoomed, 0, 255).astype(np.uint8)


# ----------------------------
# Scale factor
# ----------------------------
scale = 4

nn_img = zoom_nearest(small, scale)
bilinear_img = zoom_bilinear(small, scale)


# ----------------------------
# FIX: Resize outputs for SSD comparison
# ----------------------------
nn_img = cv2.resize(nn_img, (original.shape[1], original.shape[0]))
bilinear_img = cv2.resize(bilinear_img, (original.shape[1], original.shape[0]))


# ----------------------------
# Normalized SSD
# ----------------------------
def normalized_ssd(img1, img2):
    diff = img1.astype(np.float32) - img2.astype(np.float32)
    return np.sum(diff ** 2) / img1.size


ssd_nn = normalized_ssd(original, nn_img)
ssd_bilinear = normalized_ssd(original, bilinear_img)

print("SSD (Nearest Neighbor):", ssd_nn)
print("SSD (Bilinear):", ssd_bilinear)


# ----------------------------
# Display results in RGB format
# ----------------------------
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(original, cmap='gray')
plt.title("Original")

plt.subplot(1, 3, 2)
plt.imshow(nn_img, cmap='gray')
plt.title("Nearest Neighbor")

plt.subplot(1, 3, 3)
plt.imshow(bilinear_img, cmap='gray')
plt.title("Bilinear")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load image
img = cv2.imread(r"C:\Users\thami\Downloads\images\images\q10.jpeg")  
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Gaussian blur (similar kernel size)
gaussian = cv2.GaussianBlur(img, (9, 9), 2)

# Bilateral filter (OpenCV)
bilateral = cv2.bilateralFilter(img, d=9, sigmaColor=75, sigmaSpace=75)

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(img)
plt.title("Original")

plt.subplot(1,3,2)
plt.imshow(gaussian)
plt.title("Gaussian Blur")

plt.subplot(1,3,3)
plt.imshow(bilateral)
plt.title("Bilateral Filter")

plt.show()
```

```python

def bilateral_filter_custom(image, diameter=5, sigma_s=75, sigma_r=75):
    img = image.astype(np.float32)
    height, width, channels = img.shape
    
    pad = diameter // 2
    padded = np.pad(img, ((pad,pad),(pad,pad),(0,0)), mode='reflect')
    
    output = np.zeros_like(img)
    
    for i in range(height):
        for j in range(width):
            
            center = padded[i+pad, j+pad]
            
            weight_total = 0
            filtered = np.zeros(channels)
            
            for x in range(diameter):
                for y in range(diameter):
                    
                    neighbor = padded[i+x, j+y]
                    
                    # Spatial weight
                    distance_spatial = (x - pad)**2 + (y - pad)**2
                    spatial_weight = np.exp(-distance_spatial / (2 * sigma_s**2))
                    
                    # Range weight
                    distance_range = np.sum((neighbor - center)**2)
                    range_weight = np.exp(-distance_range / (2 * sigma_r**2))
                    
                    pixel_weight = spatial_weight * range_weight
                    
                    filtered += pixel_weight * neighbor
                    weight_total += pixel_weight
            
            output[i, j] = filtered / weight_total
    
    return np.clip(output, 0, 255).astype(np.uint8)


custom_bilateral = bilateral_filter_custom(img)

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(img)
plt.title("Original")

plt.subplot(1,3,2)
plt.imshow(bilateral)
plt.title("OpenCV Bilateral")

plt.subplot(1,3,3)
plt.imshow(custom_bilateral)
plt.title("Custom Bilateral")

plt.show()
```
