```python
import numpy as np
def intensity_transform(im, breakpoints):
    pass

breakpoints = np.array([[0, 255], [50, 200], [100, 150], [150, 100], [255, 0]])

img = plt.imread('C:\\Users\\thami\\Downloads\\images\\images\\q1.jpeg')
def intensity_transform(im, breakpoints):
    breakpoints = np.asarray(breakpoints)
    x = breakpoints[:, 0]
    y = breakpoints[:, 1]

    # Apply piecewise-linear interpolation to scalar or RGB values
    return np.interp(im, x, y)


def apply_transformation(im, breakpoints):
    return np.clip(
        intensity_transform(im, breakpoints),
        0, 255
    ).astype(np.uint8)


transformed_img = apply_transformation(img, breakpoints)

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title('Original Image')

plt.subplot(1, 2, 2)
plt.imshow(transformed_img)
plt.title('Transformed Image')

plt.show()
```

```python
img = plt.imread('C:\\Users\\thami\\Downloads\\images\\images\\q2.jpeg')
plt.imshow(img)
plt.show()
```

```python

threshold = 225
thresholded_img = (img > threshold) * 255
plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title('Original Image')

plt.subplot(1, 2, 2)
plt.imshow(thresholded_img, cmap='gray')
plt.title('White Matter Image')

plt.show()
```

```python
low_threshold = 130
high_threshold = 190
thresholded_img = ((img > low_threshold) & (img < high_threshold)) * 255
plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title('Original Image')

plt.subplot(1, 2, 2)
plt.imshow(thresholded_img, cmap='gray')
plt.title('Grey Matter Image')

plt.show()
```
